We are going to explore books with missing descriptions and see whether there are relation with book age, number of pages with missing values
- Book Age
- Number of Pages

```python
import numpy as np

books["missing_descriptions"] = np.where(books["description"].isna(), 1, 0)
books["age_of_book"] = 2025 - books["published_year"]
```

```python
import seaborn as sns
columns_of_interest = ["missing_descriptions", "age_of_book", "num_pages", "average_rating"]

corr_matrix = books[columns_of_interest].corr(method="spearman")

heatmap = sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", cbar_kws={"label": "Spearman Correlation"})

heatmap.set_title("Correlation heatmap")
plt.show()
```

```python
book_missing = books[~(books["num_pages"].isna()) &
      ~(books["description"].isna()) &
      ~(books["average_rating"].isna()) &
      ~(books["published_year"].isna())
]
```
